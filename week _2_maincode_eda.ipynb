```python
# here we will create count plots to visualize the relationship between churn and various categorical features.

# Set the visualization style
sns.set_style('whitegrid')

# Create a multi-plot layout
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# 1. Churn Class Distribution
sns.countplot(data=df, x='Churn', ax=axes[0, 0], palette='Set2')
axes[0, 0].set_title('Overall Churn Distribution')
axes[0, 0].set_ylabel('Number of Customers')

# 2. Churn rate by Contract Type
sns.countplot(data=df, x='Contract', hue='Churn', ax=axes[0, 1], palette='Set2')
axes[0, 1].set_title('Churn Distribution by Contract Type')
axes[0, 1].set_ylabel('Number of Customers')

# 3. Kernel Density Estimate of Tenure by Churn
sns.kdeplot(data=df, x='tenure', hue='Churn', fill=True, common_norm=False, ax=axes[1, 0], palette='Set2', alpha=0.5)
axes[1, 0].set_title('Customer Tenure Distribution by Churn Status')
axes[1, 0].set_xlabel('Tenure (Months)')

# 4. Kernel Density Estimate of Monthly Charges by Churn
sns.kdeplot(data=df, x='MonthlyCharges', hue='Churn', fill=True, common_norm=False, ax=axes[1, 1], palette='Set2', alpha=0.5)
axes[1, 1].set_title('Monthly Charges Distribution by Churn Status')
axes[1, 1].set_xlabel('Monthly Charges ($)')

plt.tight_layout()
plt.show()
```

```python
# here we will create count plots to visualize the relationship between churn and various categorical features.

fig, axes = plt.subplots(2, 3, figsize=(18, 11))

# 1. Gender vs Churn
sns.countplot(data=df, x='gender', hue='Churn', ax=axes[0, 0], palette='Set2')
axes[0, 0].set_title('Churn by Gender')

# 2. Senior Citizen vs Churn
sns.countplot(data=df, x='SeniorCitizen', hue='Churn', ax=axes[0, 1], palette='Set2')
axes[0, 1].set_title('Churn by Senior Citizen Status')
axes[0, 1].set_xticklabels(['No', 'Yes'])

# 3. Partner vs Churn
sns.countplot(data=df, x='Partner', hue='Churn', ax=axes[0, 2], palette='Set2')
axes[0, 2].set_title('Churn by Partner Status')

# 4. Dependents vs Churn
sns.countplot(data=df, x='Dependents', hue='Churn', ax=axes[1, 0], palette='Set2')
axes[1, 0].set_title('Churn by Dependents Status')

# 5. Internet Service vs Churn
sns.countplot(data=df, x='InternetService', hue='Churn', ax=axes[1, 1], palette='Set2')
axes[1, 1].set_title('Churn by Internet Service Type')

# 6. Payment Method vs Churn
sns.countplot(data=df, x='PaymentMethod', hue='Churn', ax=axes[1, 2], palette='Set2')
axes[1, 2].set_title('Churn by Payment Method')
axes[1, 2].set_xticklabels(axes[1, 2].get_xticklabels(), rotation=30, ha='right')

plt.tight_layout()
plt.show()
```

```python
# here we will create pie charts to visualize the proportions of churn, contract types, and internet service types.

fig, axes = plt.subplots(1, 3, figsize=(18, 6))

# 1. Churn Proportion
churn_counts = df['Churn'].value_counts()
axes[0].pie(churn_counts, labels=churn_counts.index, autopct='%1.1f%%', startangle=90, colors=['#66b3ff','#ff9999'])
axes[0].set_title('Overall Churn %')

# 2. Contract Type Proportion
contract_counts = df['Contract'].value_counts()
axes[1].pie(contract_counts, labels=contract_counts.index, autopct='%1.1f%%', startangle=90, colors=['#99ff99','#ffcc99','#c2c2f0'])
axes[1].set_title('Contract Types %')

# 3. Internet Service Proportion
internet_counts = df['InternetService'].value_counts()
axes[2].pie(internet_counts, labels=internet_counts.index, autopct='%1.1f%%', startangle=90, colors=['#ffb3e6','#c2c2f0','#ff6666'])
axes[2].set_title('Internet Service Types %')

plt.tight_layout()
plt.show()
```

```python
# here we will create pie charts to visualize the proportions of payment methods, paperless billing, and partner status.


fig, axes = plt.subplots(1, 3, figsize=(18, 6))

# 1. Payment Method Proportion
payment_counts = df['PaymentMethod'].value_counts()
axes[0].pie(payment_counts, labels=payment_counts.index, autopct='%1.1f%%', startangle=90, colors=['#ff9999','#66b3ff','#99ff99','#ffcc99'])
axes[0].set_title('Payment Methods %')

# 2. Paperless Billing Proportion
billing_counts = df['PaperlessBilling'].value_counts()
axes[1].pie(billing_counts, labels=billing_counts.index, autopct='%1.1f%%', startangle=90, colors=['#ffb3e6','#c2c2f0'])
axes[1].set_title('Paperless Billing %')

# 3. Partner Proportion
partner_counts = df['Partner'].value_counts()
axes[2].pie(partner_counts, labels=partner_counts.index, autopct='%1.1f%%', startangle=90, colors=['#ff6666','#c2f0c2'])
axes[2].set_title('Partner Status %')

plt.tight_layout()
plt.show()
```

### Key Findings and Inferences from EDA Graphs

Based on the visualizations generated above, we can draw several critical insights about customer churn:

* **Overall Churn Rate:**
  * Approximately **26.5%** of the customer base has churned, indicating a significant area of concern for customer retention.

* **Contract Types:**
  * Customers with **Month-to-month contracts** are highly prone to churning. Conversely, customers on one-year or two-year contracts show very high retention, suggesting that locking customers into longer-term contracts drastically reduces churn risk.

* **Tenure & Financials:**
  * **Tenure:** Newly acquired customers (0-10 months) have a very high concentration of churn. As tenure increases, churn behavior decreases significantly.
  * **Monthly Charges:** Churned customers are concentrated around higher monthly bill amounts (especially between $70 and $100), whereas loyal customers are mostly distributed in lower pricing tiers (around $20).

* **Demographics (Gender, Partners & Dependents):**
  * **Gender:** Churn rates are nearly identical between males and females, indicating gender is not a key driver of churn.
  * **Partners & Dependents:** Customers without a partner or dependents have a significantly higher propensity to churn compared to those with families.
  * **Senior Citizens:** Although senior citizens constitute a smaller portion of the dataset, they show a proportionally higher rate of churn compared to non-seniors.

* **Services & Billing Preferencs:**
  * **Internet Service:** Customers utilizing **Fiber optic** internet service show exceptionally high churn rates compared to DSL or those with no internet service.
  * **Paperless Billing:** Customers who opted for paperless billing have a higher rate of churn.
  * **Payment Methods:** Customers paying via **Electronic check** exhibit a disproportionately high rate of churn compared to automated payment methods (Credit Card, Bank Transfer) or mailed checks.
